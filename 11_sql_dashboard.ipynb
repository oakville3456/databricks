```python
%sql
--=============================================================
-- 11_sql_dashboard — Databricks SQL queries for dashboard
-- Author: oakville3456
-- Branch: feature/priority10-sql-dashboard
-- Purpose: Build queries to power a sales dashboard
-- =============================================================

-- Query 1: Total revenue by store
SELECT
    store_id,
    ROUND(SUM(total_revenue), 2)  AS total_revenue,
    SUM(order_count)              AS total_orders,
    SUM(unique_customers)         AS total_customers
FROM adb_retail_dev.gold.sales_daily
WHERE order_date IS NOT NULL
GROUP BY store_id
ORDER BY total_revenue DESC
```

```python
%sql
-- Query 2: Revenue trend by date
SELECT
    order_date,
    ROUND(SUM(total_revenue), 2) AS daily_revenue,
    SUM(order_count)             AS daily_orders
FROM adb_retail_dev.gold.sales_daily
WHERE order_date IS NOT NULL
GROUP BY order_date
ORDER BY order_date ASC
```

```python
%sql
-- Query 3: Top 5 products by revenue
SELECT
    product,
    ROUND(SUM(revenue), 2)  AS total_revenue,
    COUNT(order_id)         AS total_orders,
    ROUND(AVG(price), 2)    AS avg_price
FROM adb_retail_dev.silver.sales
WHERE order_date IS NOT NULL
GROUP BY product
ORDER BY total_revenue DESC
LIMIT 5
```

```python
%sql
-- Query 4: Data quality summary
SELECT
    COUNT(*)                                    AS total_rows,
    COUNT(order_id)                             AS valid_order_ids,
    SUM(CASE WHEN order_date IS NULL THEN 1 ELSE 0 END) AS null_dates,
    SUM(CASE WHEN price <= 0 THEN 1 ELSE 0 END)         AS bad_prices,
    ROUND(MIN(price), 2)                        AS min_price,
    ROUND(MAX(price), 2)                        AS max_price,
    ROUND(AVG(revenue), 2)                      AS avg_revenue,
    MIN(order_date)                             AS earliest_order,
    MAX(order_date)                             AS latest_order
FROM adb_retail_dev.silver.sales
```
